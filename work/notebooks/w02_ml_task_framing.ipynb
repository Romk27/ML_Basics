## 5. Why ML beats a fixed rule here

**ML is needed because the pattern is messy and signal interactions are complex.** A fixed rule like "flag pages with ≥20% impression drop" would miss cases where traffic is stable but engagement is quietly declining, or where keyword rankings are slipping before traffic drops. Many signals interact — CTR, position, scroll rate, content age, keyword coverage — and their relationships change across content types and over time. ML can learn these patterns from data rather than requiring a human to hand-code every edge case.
